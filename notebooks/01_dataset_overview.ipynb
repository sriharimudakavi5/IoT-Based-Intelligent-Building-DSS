```python
# =============================================================================
# SECTION 23 : DATA INTEGRATION PREPARATION
# =============================================================================
#
# Objective
# ---------
# Prepare the three research datasets for integration by converting their
# respective date and time representations into a common datetime format.
#
# The three datasets have different temporal resolutions:
#
#   • I-BLEND          → Minute-level observations
#   • NASA POWER       → Daily observations
#   • Smart Building   → Hourly observations
#
# A common datetime representation is therefore required before performing
# temporal alignment and feature integration.
#
# =============================================================================

print("=" * 90)
print("DATA INTEGRATION PREPARATION")
print("=" * 90)

print("Converting timestamp columns into datetime format...\n")
```

```python
# =============================================================================
# SECTION 24 : CONVERT I-BLEND UNIX TIMESTAMP
# =============================================================================
#
# The I-BLEND dataset stores timestamps in UNIX Epoch format.
# Convert these values into Python datetime objects and derive useful
# temporal attributes for later merging and feature engineering.
#
# =============================================================================

energy_df["datetime"] = pd.to_datetime(
    energy_df["timestamp"],
    unit="s"
)

energy_df["date"] = energy_df["datetime"].dt.date
energy_df["year"] = energy_df["datetime"].dt.year
energy_df["month"] = energy_df["datetime"].dt.month
energy_df["day"] = energy_df["datetime"].dt.day
energy_df["hour"] = energy_df["datetime"].dt.hour

print("=" * 90)
print("I-BLEND TIMESTAMP CONVERTED SUCCESSFULLY")
print("=" * 90)

display(
    energy_df[
        [
            "timestamp",
            "datetime",
            "date",
            "year",
            "month",
            "day",
            "hour"
        ]
    ].head()
)
```

```python
# =============================================================================
# SECTION 25 : CONVERT NASA DATE COLUMNS
# =============================================================================
#
# NASA POWER stores YEAR, MONTH and DAY as separate variables.
# Combine these variables into a single datetime column.
#
# =============================================================================

nasa_df["date"] = pd.to_datetime(
    nasa_df[
        ["YEAR", "MO", "DY"]
    ].rename(
        columns={
            "YEAR": "year",
            "MO": "month",
            "DY": "day"
        }
    )
)

nasa_df["date"] = nasa_df["date"].dt.date

print("=" * 90)
print("NASA DATE CREATED SUCCESSFULLY")
print("=" * 90)

display(
    nasa_df.head()
)
```

```python
# =============================================================================
# SECTION 26 : CONVERT SMART BUILDING DATETIME
# =============================================================================
#
# Convert all Smart Building timestamps into datetime format.
# These timestamps will later be used for external validation.
#
# =============================================================================

datasets = {
    "Electricity": electricity_df,
    "Weather": weather_df,
    "Cooling": cooling_df,
    "Heating": heating_df
}

for dataset_name, dataset in datasets.items():

    dataset["datetime"] = pd.to_datetime(
        dataset["datetime_utc"]
    )

    dataset["date"] = dataset["datetime"].dt.date
    dataset["year"] = dataset["datetime"].dt.year
    dataset["month"] = dataset["datetime"].dt.month
    dataset["day"] = dataset["datetime"].dt.day
    dataset["hour"] = dataset["datetime"].dt.hour

print("=" * 90)
print("SMART BUILDING TIMESTAMPS CONVERTED SUCCESSFULLY")
print("=" * 90)
```

```python
# =============================================================================
# SECTION 27 : VERIFY TEMPORAL ALIGNMENT
# =============================================================================
#
# Compare the temporal coverage of all datasets before integration.
#
# =============================================================================

summary = pd.DataFrame({

    "Dataset":[
        "I-BLEND",
        "NASA POWER",
        "Smart Building Electricity"
    ],

    "Start Date":[
        energy_df["datetime"].min(),
        nasa_df["date"].min(),
        electricity_df["datetime"].min()
    ],

    "End Date":[
        energy_df["datetime"].max(),
        nasa_df["date"].max(),
        electricity_df["datetime"].max()
    ]

})

print("=" * 90)
print("TEMPORAL COVERAGE")
print("=" * 90)

display(summary)
```

```python
# =============================================================================
# SECTION 28 : TEMPORAL RESOLUTION ANALYSIS
# =============================================================================
#
# Determine the sampling frequency of the I-BLEND dataset and verify whether
# daily aggregation is required before merging with the NASA POWER dataset.
# =============================================================================

print("=" * 100)
print("TEMPORAL RESOLUTION ANALYSIS")
print("=" * 100)

print("I-BLEND")
print(f"Minimum Timestamp : {energy_df['datetime'].min()}")
print(f"Maximum Timestamp : {energy_df['datetime'].max()}")

print("\nTotal Records")
print(len(energy_df))

print("\nUnique Dates")
print(energy_df["date"].nunique())

print("\nAverage Records Per Day")
print(round(len(energy_df) / energy_df["date"].nunique(), 2))

print("=" * 100)
```

```python
# =============================================================================
# SECTION 29 : INTEGRATE I-BLEND AND NASA POWER DATASETS
# =============================================================================
#
# Objective
# ---------
# Integrate the primary IoT building energy dataset (I-BLEND) with the NASA
# POWER environmental dataset using the common calendar date.
#
# Integration Key
# ---------------
# date
#
# Environmental Variables Added
# -----------------------------
# • ALLSKY_SFC_SW_DWN  : Solar Radiation
# • WS2M               : Wind Speed (2 m)
# • PRECTOTCORR        : Corrected Precipitation
#
# =============================================================================

# Select only the required NASA variables
nasa_features = nasa_df[
    [
        "date",
        "ALLSKY_SFC_SW_DWN",
        "WS2M",
        "PRECTOTCORR"
    ]
].copy()

# Perform left join to preserve every I-BLEND observation
integrated_df = energy_df.merge(
    nasa_features,
    how="left",
    on="date"
)

print("=" * 100)
print("DATASET INTEGRATION COMPLETED SUCCESSFULLY")
print("=" * 100)

print(f"I-BLEND Records        : {len(energy_df):,}")
print(f"NASA POWER Records     : {len(nasa_df):,}")
print(f"Integrated Records     : {len(integrated_df):,}")

print("=" * 100)
```

```python
# =============================================================================
# SECTION 30 : VERIFY DATASET INTEGRATION
# =============================================================================

print("=" * 100)
print("POST-INTEGRATION VALIDATION")
print("=" * 100)

print("\nIntegrated Dataset Shape")
print(integrated_df.shape)

print("\nNew Environmental Variables")

new_columns = [
    "ALLSKY_SFC_SW_DWN",
    "WS2M",
    "PRECTOTCORR"
]

print(integrated_df[new_columns].head())

print("\nMissing Values After Integration")

display(
    integrated_df[new_columns].isnull().sum()
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 32 : SAVE INTEGRATED DATASET
# =============================================================================

OUTPUT_DIRECTORY = PROJECT_ROOT / "data" / "processed"

OUTPUT_DIRECTORY.mkdir(exist_ok=True)

OUTPUT_FILE = OUTPUT_DIRECTORY / "integrated_building_dataset.csv"

integrated_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("=" * 100)
print("INTEGRATED DATASET SAVED SUCCESSFULLY")
print("=" * 100)

print(f"Location : {OUTPUT_FILE}")
```

```python
# =============================================================================
# SECTION 33 : MERGE SMART BUILDING DATASETS
# =============================================================================
#
# Objective
# ---------
# Combine the Smart Building electricity, weather, cooling, and heating
# datasets into a single unified dataset using the common timestamp.
#
# This unified dataset will be used for external validation of the
# proposed Intelligent Building Decision Support System.
#
# =============================================================================

# Merge Electricity and Weather
smart_building_df = electricity_df.merge(
    weather_df,
    on="datetime_utc",
    how="outer"
)

# Merge Cooling
smart_building_df = smart_building_df.merge(
    cooling_df,
    on="datetime_utc",
    how="outer",
    suffixes=("", "_cooling")
)

# Merge Heating
smart_building_df = smart_building_df.merge(
    heating_df,
    on="datetime_utc",
    how="outer",
    suffixes=("", "_heating")
)

print("=" * 100)
print("SMART BUILDING DATASETS MERGED SUCCESSFULLY")
print("=" * 100)

print(f"Rows    : {smart_building_df.shape[0]:,}")
print(f"Columns : {smart_building_df.shape[1]}")

display(smart_building_df.head())
```

```python
# =============================================================================
# SECTION 34 : VALIDATE SMART BUILDING DATASET
# =============================================================================

print("=" * 100)
print("SMART BUILDING DATASET VALIDATION")
print("=" * 100)

print(f"Shape : {smart_building_df.shape}")

print("\nMissing Values")
display(smart_building_df.isnull().sum())

print("\nDuplicate Records")
print(smart_building_df.duplicated().sum())

print("\nColumn Names")
print(list(smart_building_df.columns))
```

```python
# =============================================================================
# SECTION 35 : CREATE CLEAN SMART BUILDING VALIDATION DATASET
# =============================================================================
#
# Objective
# ---------
# Create a clean validation dataset by removing duplicate temporal columns
# generated during dataset merging and assigning meaningful feature names.
#
# =============================================================================

smart_validation_df = smart_building_df.copy()

# Remove duplicate temporal columns
columns_to_drop = [
    "datetime_x", "date_x", "year_x", "month_x", "day_x", "hour_x",
    "datetime_y", "date_y", "year_y", "month_y", "day_y", "hour_y",
    "datetime_heating", "date_heating", "year_heating",
    "month_heating", "day_heating", "hour_heating"
]

smart_validation_df.drop(
    columns=columns_to_drop,
    inplace=True,
    errors="ignore"
)

# Rename columns for clarity
smart_validation_df.rename(columns={
    "total": "electricity_total",
    "WeatherStation.Weather.Igm": "solar_irradiance",
    "WeatherStation.Weather.Ta": "air_temperature"
}, inplace=True)

print("=" * 100)
print("CLEAN SMART BUILDING VALIDATION DATASET CREATED")
print("=" * 100)

print(f"Rows    : {smart_validation_df.shape[0]:,}")
print(f"Columns : {smart_validation_df.shape[1]}")

print("\nColumns:")
print(list(smart_validation_df.columns))
```

```python
# =============================================================================
# SECTION 36 : SAVE SMART BUILDING VALIDATION DATASET
# =============================================================================

VALIDATION_FILE = OUTPUT_DIRECTORY / "smart_building_validation_dataset.csv"

smart_validation_df.to_csv(
    VALIDATION_FILE,
    index=False
)

print("=" * 100)
print("SMART BUILDING VALIDATION DATASET SAVED")
print("=" * 100)

print(f"Location : {VALIDATION_FILE}")
```
